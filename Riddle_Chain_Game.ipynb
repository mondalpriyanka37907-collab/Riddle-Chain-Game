```python
import random
import matplotlib.pyplot as plt
from collections import Counter

def word_to_number(word):
    word = word.lower()
    total = 0
    for ch in word:
        if 'a' <= ch <= 'z':  # only letters
            total += (ord(ch) - ord('a') + 1)
    return total

def riddle_with_frequency():
    riddles = [
        {"question": "I speak without a mouth and hear without ears. What am I?", "answer": "echo"},
        {"question": "The more you take, the more you leave behind. What am I?", "answer": "footsteps"},
        {"question": "What has to be broken before you can use it?", "answer": "egg"},
        {"question": "I’m tall when I’m young, and short when I’m old. What am I?", "answer": "candle"}
    ]

    random.shuffle(riddles)
    riddle = riddles[0]
    print("\n Riddle Challenge")
    print(riddle["question"])
    user_answer = input("Your answer: ").lower().strip()
    if user_answer == riddle["answer"]:
        print(f" Correct! The answer is {user_answer}\n You may continue to the next challenge.")
        text = riddle["question"].lower()
        for ch in ['.', ',', '!', '?', ';', ':', '-', '_', '"', "'"]:
            text = text.replace(ch, "")
        words = text.split()
        word_counts = Counter(words)
        print("\n Word Frequencies in the riddle:")
        for word, count in word_counts.most_common(5):
            print(f"{word}: {count}")
        top_words, top_counts = zip(*word_counts.most_common(5))
        plt.bar(top_words, top_counts, color="skyblue")
        plt.xticks(rotation=30)
        plt.title("Word Frequency in Riddle")
        plt.xlabel("Words")
        plt.ylabel("Frequency")
        plt.yticks(range(1, max(top_counts) + 1))
        plt.tight_layout()
        plt.show()
        plt.close()

        most_common_word, count = word_counts.most_common(1)[0]
        print(f"\n Clue: The answer of the riddle '{riddle['answer']}'.")
        secret_number = word_to_number(riddle["answer"])
        return secret_number
    else:
        print(f" Wrong! The correct answer was '{riddle['answer']}'. Game Over.")
        return None

def number_guessing_game(secret_number):
    print("\n Number Guessing Challenge")
    print("Now guess the secret number derived from the riddle answer in 5 attempts!")

    attempts = 5
    for attempt in range(1, attempts + 1):
        try:
            guess = int(input(f"Attempt {attempt}/{attempts} - Enter your guess: "))
        except ValueError:
            print(" Invalid input! Please enter a number.")
            continue

        if guess == secret_number:
            print(f" Correct! You solved the full riddle chain in {attempt} attempts!")
            return True
        elif guess < secret_number:
            print("Too low!")
        else:
            print("Too high!")
    else:
        print(f" Game Over! The secret number was {secret_number}")
        return False

def main_game():
    print("\n Welcome to the Riddle Chain Game!")
    print("Step 1: Solve the riddle.")
    print("Step 2: The Word Frequency Analyzer reveals a clue.")
    print("Step 3: The riddle answer is converted into a secret number (sum of letters).")
    print("Step 4: Guess that number!")

    secret_number = riddle_with_frequency()
    if secret_number:
        number_guessing_game(secret_number)
    else:
        print(" You failed the riddle, so the chain ends here.")

if __name__ == "__main__":
    main_game()
```
